# Mindestanforderungen 1 - NumPy und pandas

```python
import csv
import sys

import numpy as np
import pandas as pd
```

Erstellen Sie NumPy-Arrays mit (Teil-)Daten aus Ihrem Projekt - es sollten verschiedene Datentypen (Ganzzahl, Kommazahl, String, Boolean, etc.) vorhanden sein

```python
rows = []

csv.field_size_limit(sys.maxsize)

with open("survey_results_public.csv", "r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)
    cleaned = []
    for name in reader.fieldnames:
        cleaned.append(name.strip().replace('"', ''))
    reader.fieldnames = cleaned

    for row in reader:
        try:
            rows.append([
                int(row["ResponseId"]),
                row["MainBranch"],
                row["Age"],
                str(row["EdLevel"]),
                row["Country"],
                float(row["WorkExp"]) if row["WorkExp"] != "NA" else np.nan,
                float(row["CompTotal"]) if row["CompTotal"] != "NA" else np.nan,
            ])
        except ValueError:
            continue  # Zeilen mit fehlerhaften Werten überspringen

# In NumPy-Array umwandeln
data = np.array(rows, dtype=object)

print("Shape:", data.shape)
print("Beispielzeile:", data[0])
```

```python
response_id = data[:,0]
main_branch = data[:,1]
age = data[:,2]
ed_level = data[:,3]
country = data[:,4]
work_exp = data[:,5]
comp_total = data[:,6]

print(response_id)
print(main_branch)
print(age)
print(ed_level)
print(country)
print(work_exp)
print(comp_total)
```

Filtern Sie ein NumPy-Array basierend auf mindestens 2 Bedingungen auf mindestens 2 Arrays/Spalten

```python
comp_total = pd.to_numeric(comp_total)
work_exp   = pd.to_numeric(work_exp)

mask = (comp_total > 60000.0) & (work_exp > 7.0)
filtered_rows = data[mask]

print("Treffer:", filtered_rows.shape[0])
print(filtered_rows[:5])
```

Berechnen Sie mind. 2 Statistiken auf mind. 2 Arrays

```python
comp_total_mean = round(np.nanmean(comp_total),2)
print("Durchschnittliches Einkommen (unterschiedliche Währungen):",comp_total_mean)

work_exp_mean = round(np.nanmean(work_exp),2)
print("Durchschnittliche Arbeitserfahrung:", work_exp_mean, "Jahre")

comp_total_median = round(np.nanmedian(comp_total),2)
print("Median Einkommen:", comp_total_median)

work_exp_median = round(np.nanmedian(work_exp),2)
print("Median Arbeitserfahrung:", work_exp_median)
```

Laden Sie eine Datei von der Festplatte oder einer URL als pandas-DataFrame

```python
df = pd.read_csv('survey_results_public.csv')

df.set_index("ResponseId", inplace=True)
df.head(10)
```

Fügen Sie mindestens eine neue Spalte hinzu, die aus mindestens einer anderen Spalte berechnet wurde

```python
pos = df.columns.get_loc("ConvertedCompYearly") + 1
monthly = df["ConvertedCompYearly"]/12

df.insert(pos, "ConvertedCompMonthly", monthly)
```

Ändern Sie Reihenfolge und Benennung der Spalten

```python
df = df.rename(columns={
    "JobSat": "JobSatisfaction"
})
```

```python
col_country = df.pop("Country")
df.insert(4, "Country", col_country)
```

Löschen Sie mindestens eine Spalte

```python
if "AIAgentObsWrite" in df.columns:
    df.drop("AIAgentObsWrite", axis=1, inplace=True)
```

Verwenden Sie mindestens eine Funtion aus dem *str*-Accessor

```python
df["EdLevel"] = df["EdLevel"].str.split("(").str[0].str.strip()
```
