```python
rows = []

csv.field_size_limit(sys.maxsize)

with open("survey_results_public.csv", "r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)
    cleaned = []
    for name in reader.fieldnames:
        cleaned.append(name.strip().replace('"', ''))
    reader.fieldnames = cleaned

    for row in reader:
        try:
            rows.append([
                int(row["ResponseId"]),                     # Ganzzahl
                row["MainBranch"],
                row["Age"],
                str(row["EdLevel"]),
                row["Country"],                             # String
                float(row["WorkExp"]) if row["WorkExp"] != "NA" else np.nan,  # Kommazahl
                float(row["CompTotal"]) if row["CompTotal"] != "NA" else np.nan,  # Kommazahl
            ])
        except ValueError:
            continue  # Zeilen mit fehlerhaften Werten überspringen

# In NumPy-Array umwandeln
data = np.array(rows, dtype=object)

print("Shape:", data.shape)
print("Beispielzeile:", data[0])
```

```python
response_id = data[:,0]
main_branch = data[:,1]
age = data[:,2]
ed_level = data[:,3]
country = data[:,4]
work_exp = data[:,5]
comp_total = data[:,6]

print(response_id)
print(main_branch)
print(age)
print(ed_level)
print(country)
print(work_exp)
print(comp_total)
```

Filtern Sie ein NumPy-Array basierend auf mindestens 2 Bedingungen auf mindestens 2 Arrays/Spalten

```python
filt_comp_total = comp_total > 60000.0
filt_work_exp = work_exp > 7.0

mask = filt_comp_total & filt_work_exp
filtered_rows = data[mask]

print("Treffer:", filtered_rows.shape[0])
print(filtered_rows[:5])
```

Berechnen Sie mind. 2 Statistiken auf mind. 2 Arrays

```python
comp_total_mean = round(np.nanmean(comp_total),2)
print("Durchschnittliches Einkommen (unterschiedliche Währungen):",comp_total_mean)

work_exp_mean = round(np.nanmean(work_exp),2)
print("Durchschnittliche Arbeitserfahrung:", work_exp_mean, "Jahre")

comp_total_median = round(np.nanmedian(comp_total),2)
print("Median Einkommen:", comp_total_median)

work_exp_median = round(np.nanmedian(work_exp),2)
print("Median Arbeitserfahrung:", work_exp_median)
```
